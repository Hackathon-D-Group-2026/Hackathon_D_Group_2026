# Town Flu Prevention and Cost Minimization

This notebook aims to minimize the town's total expenditure, which consists of:
1. **Vaccination Costs**: Cost of buying and delivering vaccines to those who accept them, plus compensation for those who wanted a vaccine but did not receive one.
2. **Hospitalization Costs**: The average number of hospitalized people multiplied by the average hospitalization cost.

### Assumptions & Constants:
* **Infection Probability**: 100% within indoor parties.
* **Vaccine Supply Constraint**: Vaccines must be exactly matched to willing recipients to avoid ruin.
* **Compensation**: Anyone wanting a vaccine who doesn't get one is reimbursed the vaccine cost.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
import lightgbm as lgb
import warnings
warnings.filterwarnings('ignore')

# Prepare features
cat_cols = ['age_group', 'education', 'race', 'sex', 'income_poverty', 'marital_status',
            'rent_or_own', 'employment_status', 'census_msa', 'employment_sector']

# Combine for unified processing
train_df['is_train'] = 1
test_df['is_train'] = 0
test_df['seasonal_vaccine'] = np.nan
all_df = pd.concat([train_df, test_df], ignore_index=True)

# Encode categorical variables
for col in cat_cols:
    all_df[col] = all_df[col].astype('category')

# Split back
train_processed = all_df[all_df['is_train'] == 1].copy()
test_processed = all_df[all_df['is_train'] == 0].copy()

features = [c for c in train_processed.columns if c not in ['respondent_id', 'seasonal_vaccine', 'is_train']]
target = 'seasonal_vaccine'

# Train LightGBM model using cross-validation
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof = np.zeros(len(train_processed))
test_preds = np.zeros(len(test_processed))

for fold, (train_idx, val_idx) in enumerate(kf.split(train_processed, train_processed[target])):
    X_tr, y_tr = train_processed.iloc[train_idx][features], train_processed.iloc[train_idx][target]
    X_va, y_va = train_processed.iloc[val_idx][features], train_processed.iloc[val_idx][target]

    model = lgb.LGBMClassifier(
        n_estimators=1000,
        learning_rate=0.03,
        num_leaves=31,
        random_state=42 + fold,
        verbose=-1
    )
    model.fit(
        X_tr, y_tr,
        eval_set=[(X_va, y_va)],
        callbacks=[lgb.early_stopping(50, verbose=False)]
    )

    oof[val_idx] = model.predict_proba(X_va)[:, 1]
    test_preds += model.predict_proba(test_processed[features])[:, 1] / 5

print("OOF AUC Score:", roc_auc_score(train_processed[target], oof))
```

```python
# Define risk group for each individual
def is_high_risk(row):
    age_high = row['age_group'] == '65+ Years'
    # Convert chronic conditions to numeric, handling missing values by setting them to 0
    try:
        chronic_val = float(row['number_chronic_med_conditions'])
    except:
        chronic_val = 0.0
    chronic_high = chronic_val > 0
    return int(age_high or chronic_high)

# Let's map high-risk and calculate optimization on the test set
test_processed['high_risk'] = test_processed.apply(is_high_risk, axis=1)
test_processed['p_vacc'] = test_preds

# Cost parameters from PDF and instructions
C_V = 300.0
C_H = 18500.0

# High risk probabilities
# P(H | high-risk, unvaccinated) = 0.041
# P(H | high-risk, vaccinated) = 0.041 * 0.5 = 0.0205
# Low risk probabilities
# P(H | low-risk, unvaccinated) = 0.011
# P(H | low-risk, vaccinated) = 0.011 * 0.5 = 0.0055

def determine_optimal_decision(row):
    p = row['p_vacc']
    is_hr = row['high_risk']

    if is_hr == 1:
        p_h_unvax = 0.041
        p_h_vax = 0.0205
    else:
        p_h_unvax = 0.011
        p_h_vax = 0.0055

    # Case 1: We decide to offer the vaccine (Decision = 1)
    # If citizen accepts (prob p): cost is C_V + p_h_vax * C_H
    # If citizen rejects (prob 1-p): cost is p_h_unvax * C_H (ruined vaccine cost is not paid, but they are not compensated since we offered)
    expected_cost_offer = p * (C_V + p_h_vax * C_H) + (1.0 - p) * (p_h_unvax * C_H)

    # Case 2: We decide NOT to offer the vaccine (Decision = 0)
    # If citizen accepts/wants (prob p): we must pay compensation C_V, and they are unvaccinated: C_V + p_h_unvax * C_H
    # If citizen rejects (prob 1-p): cost is just hospitalization cost: p_h_unvax * C_H
    expected_cost_no_offer = p * (C_V + p_h_unvax * C_H) + (1.0 - p) * (p_h_unvax * C_H)

    # Simplify expected_cost_no_offer: p * C_V + p_h_unvax * C_H
    # Compare and return 1 (offer) or 0 (no offer)
    if expected_cost_offer < expected_cost_no_offer:
        return 1.0
    else:
        return 0.0

test_processed['seasonal_vaccine'] = test_processed.apply(determine_optimal_decision, axis=1)

# Create submission dataframe
submission = test_processed[['respondent_id', 'seasonal_vaccine']].copy()
submission['seasonal_vaccine'] = submission['seasonal_vaccine'].astype(float)
submission.to_csv('/content/submission.csv', index=False)

print("Submission successfully created and saved at /content/submission.csv!")
print(submission.head(10))
```

```python
# Map high-risk and calculate expected cost on the training/validation out-of-fold predictions
train_processed['high_risk'] = train_processed.apply(is_high_risk, axis=1)
train_processed['p_vacc'] = oof

# Calculate expected cost for each individual using our decision rule
def calculate_min_expected_cost(row):
    p = row['p_vacc']
    is_hr = row['high_risk']

    if is_hr == 1:
        p_h_unvax = 0.041
        p_h_vax = 0.0205
    else:
        p_h_unvax = 0.011
        p_h_vax = 0.0055

    expected_cost_offer = p * (C_V + p_h_vax * C_H) + (1.0 - p) * (p_h_unvax * C_H)
    expected_cost_no_offer = p * (C_V + p_h_unvax * C_H) + (1.0 - p) * (p_h_unvax * C_H)

    return min(expected_cost_offer, expected_cost_no_offer)

train_processed['expected_cost'] = train_processed.apply(calculate_min_expected_cost, axis=1)
total_oof_expected_cost = train_processed['expected_cost'].sum()
mean_oof_expected_cost = train_processed['expected_cost'].mean()

print(f"Total OOF Expected Cost for the town ({len(train_processed)} residents): ${total_oof_expected_cost:,.2f}")
print(f"Average Expected Cost per resident: ${mean_oof_expected_cost:,.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate expected hospitalization cost for each individual in train_processed
# Expected hospitalization cost with the optimal choice of offering a vaccine (which we do for 100% since it is optimal):
# If they want it (prob p): they get vaccinated -> p_h_vax * C_H
# If they do not want it (prob 1-p): they stay unvaccinated -> p_h_unvax * C_H

def get_expected_hospitalization_cost(row):
    p = row['p_vacc']
    is_hr = row['high_risk']

    if is_hr == 1:
        p_h_unvax = 0.041
        p_h_vax = 0.0205
    else:
        p_h_unvax = 0.011
        p_h_vax = 0.0055

    # Under optimal policy (offer=1):
    return (p * p_h_vax + (1 - p) * p_h_unvax) * C_H

train_processed['expected_hospitalization_cost'] = train_processed.apply(get_expected_hospitalization_cost, axis=1)

# Create the visualization
plt.figure(figsize=(10, 6))
sns.histplot(
    data=train_processed,
    x='expected_hospitalization_cost',
    hue='high_risk',
    kde=True,
    bins=30,
    multiple='stack',
    palette='viridis'
)

plt.title('Distribution of Expected Hospitalization Costs: High-Risk vs Low-Risk')
plt.xlabel('Expected Hospitalization Cost ($)')
plt.ylabel('Count')
plt.legend(title='Group Status', labels=['High-Risk (Elderly or Chronic)', 'Low-Risk'])
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.base import clone
from sklearn.model_selection import (
    StratifiedKFold, RandomizedSearchCV
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder, StandardScaler
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report, roc_curve
)

from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

from scipy.stats import randint, uniform, loguniform

# ------------------------------------------------
# 1. Prepare data
# ------------------------------------------------

target = 'seasonal_vaccine'

excluded = [
    'respondent_id', target, 'is_train',
    'high_risk', 'p_vacc', 'expected_cost',
    'expected_hospitalization_cost'
]

features = [c for c in train_df.columns
            if c not in excluded and c in test_df.columns]

X = train_df[features].copy()
X_test = test_df[features].copy()
y = train_df[target].astype(int).values

cat_features = [
    c for c in features
    if c in cat_cols or
    str(X[c].dtype) in ['object', 'category', 'string']
]

num_features = [
    c for c in features if c not in cat_features
]

for col in cat_features:
    X[col] = X[col].astype(object)
    X_test[col] = X_test[col].astype(object)

# Dense one-hot encoding for compatibility across models
try:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse=False
    )

preprocessor = ColumnTransformer(
    transformers=[
        ('numeric', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), num_features),

        ('categorical', Pipeline([
            ('imputer', SimpleImputer(
                strategy='constant',
                fill_value='Missing'
            )),
            ('encoder', encoder)
        ]), cat_features)
    ],
    sparse_threshold=0
)

# ------------------------------------------------
# 2. Define five models and search spaces
# ------------------------------------------------

model_configs = {
    'LightGBM': (
        LGBMClassifier(
            random_state=42, verbose=-1,
            n_jobs=1
        ),
        {
            'model__n_estimators': [300, 500, 800],
            'model__learning_rate': [0.02, 0.03, 0.05],
            'model__num_leaves': [15, 31, 50],
            'model__max_depth': [-1, 8, 12]
        }
    ),

    'Random Forest': (
        RandomForestClassifier(
            random_state=42, n_jobs=1
        ),
        {
            'model__n_estimators': [200, 300, 500],
            'model__max_depth': [10, 15, 20, None],
            'model__min_samples_leaf': [1, 2, 4],
            'model__max_features': ['sqrt', 0.8]
        }
    ),

    'XGBoost': (
        XGBClassifier(
            eval_metric='logloss',
            random_state=42,
            n_jobs=1
        ),
        {
            'model__n_estimators': [300, 500],
            'model__max_depth': [3, 4, 5],
            'model__learning_rate': [0.03, 0.05, 0.1],
            'model__subsample': [0.8, 1.0],
            'model__colsample_bytree': [0.8, 1.0]
        }
    ),

    'CatBoost': (
        CatBoostClassifier(
            verbose=False,
            random_seed=42,
            allow_writing_files=False,
            thread_count=1
        ),
        {
            'model__iterations': [300, 500],
            'model__depth': [4, 6, 8],
            'model__learning_rate': [0.03, 0.05, 0.1]
        }
    ),

    'Logistic Regression': (
        LogisticRegression(
            max_iter=2000,
            random_state=42
        ),
        {
            'model__C': [0.01, 0.1, 1, 10],
            'model__solver': ['lbfgs']
        }
    )
}

# ------------------------------------------------
# 3. Nested cross-validation
# ------------------------------------------------

outer_cv = StratifiedKFold(
    n_splits=5, shuffle=True, random_state=42
)

model_oof = {}
best_parameters = {}

for name, (classifier, parameters) in model_configs.items():

    print(f"\nStarting nested CV: {name}")

    oof_predictions = np.zeros(len(X))
    fold_scores = []
    fold_params = []

    for fold, (tr_idx, val_idx) in enumerate(
        outer_cv.split(X, y), 1
    ):

        X_tr = X.iloc[tr_idx]
        X_val = X.iloc[val_idx]
        y_tr = y[tr_idx]

        pipeline = Pipeline([
            ('preprocessing', clone(preprocessor)),
            ('model', clone(classifier))
        ])

        inner_cv = StratifiedKFold(
            n_splits=3,
            shuffle=True,
            random_state=42 + fold
        )

        search = RandomizedSearchCV(
            estimator=pipeline,
            param_distributions=parameters,
            n_iter=3,
            scoring='roc_auc',
            cv=inner_cv,
            random_state=42,
            n_jobs=1,
            refit=True
        )

        search.fit(X_tr, y_tr)

        probabilities = search.best_estimator_.predict_proba(
            X_val
        )[:, 1]

        oof_predictions[val_idx] = probabilities

        fold_auc = roc_auc_score(
            y[val_idx], probabilities
        )

        fold_scores.append(fold_auc)
        fold_params.append(search.best_params_)

        print(
            f"Fold {fold}: AUC = {fold_auc:.4f}"
        )

    model_oof[name] = oof_predictions
    best_parameters[name] = fold_params

    print(
        f"{name} Mean Fold AUC:",
        round(np.mean(fold_scores), 4)
    )

print("\nNested cross-validation completed.")
```

```python

# ------------------------------------------------
# Cost assumptions from your project
# ------------------------------------------------

C_V = 300.0
C_H = 18500.0

P_HIGH_UNVAX = 0.041
P_HIGH_VAX = 0.0205

P_LOW_UNVAX = 0.011
P_LOW_VAX = 0.0055

# Identify high-risk residents
def identify_high_risk(df):

    age_high = (
        df['age_group'].astype(str) == '65+ Years'
    )

    chronic = pd.to_numeric(
        df['number_chronic_med_conditions'],
        errors='coerce'
    ).fillna(0)

    return (
        age_high | (chronic > 0)
    ).astype(int).values


risk = identify_high_risk(train_df)

def calculate_cost(probabilities, risk):

    total_cost = 0
    offer_cost_total = 0
    no_offer_cost_total = 0
    offers = 0

    for p, hr in zip(probabilities, risk):

        if hr == 1:
            p_unvax = P_HIGH_UNVAX
            p_vax = P_HIGH_VAX
        else:
            p_unvax = P_LOW_UNVAX
            p_vax = P_LOW_VAX

        # Offer vaccine
        cost_offer = (
            p * (C_V + p_vax * C_H)
            + (1-p) * p_unvax * C_H
        )

        # Do not offer vaccine
        cost_no_offer = (
            p * (C_V + p_unvax * C_H)
            + (1-p) * p_unvax * C_H
        )

        # Optimal decision
        if cost_offer < cost_no_offer:
            total_cost += cost_offer
            offers += 1
        else:
            total_cost += cost_no_offer

        offer_cost_total += cost_offer
        no_offer_cost_total += cost_no_offer

    return {
        'Expected Total Cost': total_cost,
        'Average Cost per Resident': total_cost / len(probabilities),
        'Vaccine Offers': offers,
        'Offer Only Cost': offer_cost_total,
        'No Offer Only Cost': no_offer_cost_total
    }


# ------------------------------------------------
# Evaluate all models
# ------------------------------------------------

cost_results = []

for name, probabilities in model_oof.items():

    cost = calculate_cost(probabilities, risk)

    cost_results.append({
        'Model': name,
        **cost
    })

cost_df = pd.DataFrame(cost_results)

print("\nCOST OPTIMIZATION RESULTS")
print(cost_df.round(2).to_string(index=False))

# ------------------------------------------------
# Combined classification and cost metrics
# ------------------------------------------------

final_results = []

for name, probabilities in model_oof.items():

    predictions = (
        probabilities >= 0.5
    ).astype(int)

    cost = calculate_cost(probabilities, risk)

    final_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y, predictions),
        'Precision': precision_score(y, predictions),
        'Recall': recall_score(y, predictions),
        'F1 Score': f1_score(y, predictions),
        'ROC AUC': roc_auc_score(y, probabilities),
        'Expected Total Cost': cost['Expected Total Cost'],
        'Average Cost': cost['Average Cost per Resident']
    })

final_results_df = pd.DataFrame(final_results)

print("\nFINAL MODEL EVALUATION")
print(final_results_df.round(4).to_string(index=False))
```

```python

import os
import joblib

os.makedirs('/content/hackathon_submissions', exist_ok=True)

final_test_predictions = {}
submission_files = []

for name, (classifier, parameters) in model_configs.items():

    print(f"\nFinal training: {name}")

    pipeline = Pipeline([
        ('preprocessing', clone(preprocessor)),
        ('model', clone(classifier))
    ])

    search = RandomizedSearchCV(
        estimator=pipeline,
        param_distributions=parameters,
        n_iter=5,
        scoring='roc_auc',
        cv=StratifiedKFold(
            n_splits=3,
            shuffle=True,
            random_state=42
        ),
        random_state=42,
        n_jobs=1,
        refit=True
    )

    search.fit(X, y)

    model = search.best_estimator_

    test_probabilities = model.predict_proba(
        X_test
    )[:, 1]

    final_test_predictions[name] = test_probabilities

    # Save probability predictions
    probability_submission = pd.DataFrame({
        'respondent_id': test_df['respondent_id'],
        'seasonal_vaccine': test_probabilities
    })

    safe_name = name.lower().replace(' ', '_')

    probability_path = (
        f'/content/hackathon_submissions/'
        f'{safe_name}_probabilities.csv'
    )

    probability_submission.to_csv(
        probability_path, index=False
    )

    # Also save economic policy decisions
    test_risk = identify_high_risk(test_df)

    test_cost = calculate_cost(
        test_probabilities, test_risk
    )

    decisions = []

    for p, hr in zip(test_probabilities, test_risk):

        if hr:
            pu, pv = P_HIGH_UNVAX, P_HIGH_VAX
        else:
            pu, pv = P_LOW_UNVAX, P_LOW_VAX

        offer = p * (C_V + pv * C_H) + (1-p) * pu * C_H
        no_offer = p * (C_V + pu * C_H) + (1-p) * pu * C_H

        decisions.append(int(offer < no_offer))

    decision_submission = pd.DataFrame({
        'respondent_id': test_df['respondent_id'],
        'seasonal_vaccine': decisions
    })

    decision_path = (
        f'/content/hackathon_submissions/'
        f'{safe_name}_decisions.csv'
    )

    decision_submission.to_csv(
        decision_path, index=False
    )

    # Save final fitted model
    joblib.dump(
        model,
        f'/content/hackathon_submissions/{safe_name}_model.pkl'
    )

    submission_files.append(probability_path)

    print("Best parameters:", search.best_params_)
    print("Best tuning CV AUC:", round(search.best_score_, 4))
    print("Test expected cost:", round(
        test_cost['Expected Total Cost'], 2
    ))

print("\nAll five model submissions generated!")
```
