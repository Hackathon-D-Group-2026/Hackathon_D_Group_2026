```python

# CELL 1: IMPORT LIBRARIES AND LOAD DATA

!pip install -q lightgbm xgboost catboost openpyxl

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import lightgbm as lgb
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import (
    StratifiedKFold, RandomizedSearchCV
)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report, roc_curve
)
from sklearn.calibration import calibration_curve

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')

# Load datasets
train_df = pd.read_csv('/content/dataset_A_training.csv')
test_df = pd.read_csv('/content/dataset_A_testing.csv')
submission_df = pd.read_csv('/content/dataset_A_example_submission.csv')

TARGET = 'seasonal_vaccine'

print("Training:", train_df.shape)
print("Testing:", test_df.shape)
print("Submission:", submission_df.shape)

display(train_df.head())
```

```python

# CELL 2: EXPLORATORY DATA ANALYSIS

print("DATASET INFORMATION")
train_df.info()

print("\nMISSING VALUES")
display(
    train_df.isnull().sum()
    .sort_values(ascending=False)
    .to_frame('Missing Values')
)

print("\nTARGET DISTRIBUTION")
display(train_df[TARGET].value_counts(normalize=True))

# Missing values visualization
missing = train_df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if len(missing):
    plt.figure(figsize=(10,5))
    missing.plot(kind='bar', color='coral')
    plt.title('Missing Values Analysis')
    plt.tight_layout()
    plt.show()

# Target distribution
fig, axes = plt.subplots(1, 2, figsize=(13,5))

sns.countplot(
    data=train_df, x=TARGET,
    hue=TARGET, palette='Set2',
    legend=False, ax=axes[0]
)
axes[0].set_title('Vaccination Class Distribution')

train_df.select_dtypes(include=np.number).hist(
    figsize=(14,8), bins=25
)
plt.suptitle('Numerical Feature Distributions')
plt.tight_layout()
plt.show()

# Correlation
numeric_data = train_df.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))
sns.heatmap(
    numeric_data.corr(),
    cmap='coolwarm',
    center=0
)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()

# Categorical features
cat_cols = [
    'age_group', 'education', 'race', 'sex',
    'income_poverty', 'marital_status',
    'rent_or_own', 'employment_status',
    'census_msa', 'employment_sector'
]

cat_cols = [c for c in cat_cols if c in train_df.columns]

for col in cat_cols:
    plt.figure(figsize=(9,4))
    train_df[col].value_counts().head(12).plot(
        kind='bar', color='steelblue'
    )
    plt.title(f'Distribution: {col}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

print("EDA completed.")
```

```python

# CELL 3: DATA PREPROCESSING

TARGET = 'seasonal_vaccine'

excluded = [
    'respondent_id', TARGET, 'is_train',
    'high_risk', 'p_vacc', 'expected_cost',
    'expected_hospitalization_cost'
]

features = [
    c for c in train_df.columns
    if c not in excluded and c in test_df.columns
]

X = train_df[features].copy()
X_test = test_df[features].copy()
y = train_df[TARGET].astype(int).values

cat_features = [
    c for c in features
    if c in cat_cols or
    str(X[c].dtype) in ['object', 'category', 'string']
]

num_features = [
    c for c in features if c not in cat_features
]

for col in cat_features:
    X[col] = X[col].astype(object)
    X_test[col] = X_test[col].astype(object)

# Numerical and categorical processing
try:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown='ignore',
        sparse=False
    )

preprocessor = ColumnTransformer([
    ('numeric', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), num_features),

    ('categorical', Pipeline([
        ('imputer', SimpleImputer(
            strategy='constant',
            fill_value='Missing'
        )),
        ('encoder', encoder)
    ]), cat_features)
], sparse_threshold=0)

print("Numerical features:", len(num_features))
print("Categorical features:", len(cat_features))
print("Total features:", len(features))
print("Preprocessing pipeline created.")
```

```python

# CELL 4: MODEL DEFINITIONS AND HYPERPARAMETER SPACES

models = {
    'LightGBM': (
        lgb.LGBMClassifier(
            random_state=42,
            verbose=-1,
            n_jobs=1
        ),
        {
            'model__n_estimators': [300, 500, 800],
            'model__learning_rate': [0.02, 0.03, 0.05],
            'model__num_leaves': [15, 31, 50],
            'model__max_depth': [-1, 8, 12]
        }
    ),

    'Random Forest': (
        RandomForestClassifier(
            random_state=42,
            n_jobs=1
        ),
        {
            'model__n_estimators': [200, 300, 500],
            'model__max_depth': [10, 15, 20, None],
            'model__min_samples_leaf': [1, 2, 4]
        }
    ),

    'XGBoost': (
        XGBClassifier(
            eval_metric='logloss',
            random_state=42,
            n_jobs=1
        ),
        {
            'model__n_estimators': [300, 500],
            'model__max_depth': [3, 4, 5],
            'model__learning_rate': [0.03, 0.05, 0.1],
            'model__subsample': [0.8, 1.0]
        }
    ),

    'CatBoost': (
        CatBoostClassifier(
            iterations=500,
            verbose=False,
            random_seed=42,
            allow_writing_files=False,
            thread_count=1
        ),
        {
            'model__iterations': [300, 500],
            'model__depth': [4, 6, 8],
            'model__learning_rate': [0.03, 0.05, 0.1]
        }
    ),

    'Logistic Regression': (
        LogisticRegression(
            max_iter=2000,
            random_state=42
        ),
        {
            'model__C': [0.01, 0.1, 1, 10]
        }
    )
}

print("Five models successfully initialized.")
```

```python

# CELL 5: NESTED CROSS VALIDATION AND TUNING

from sklearn.base import clone

outer_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

model_oof = {}
best_parameters = {}

for name, (classifier, parameters) in models.items():

    print(f"\nTraining {name}")

    predictions = np.zeros(len(X))
    fold_scores = []
    best_params = []

    for fold, (tr_idx, val_idx) in enumerate(
        outer_cv.split(X, y), 1
    ):

        pipeline = Pipeline([
            ('preprocessing', clone(preprocessor)),
            ('model', clone(classifier))
        ])

        inner_cv = StratifiedKFold(
            n_splits=3,
            shuffle=True,
            random_state=42 + fold
        )

        search = RandomizedSearchCV(
            pipeline,
            parameters,
            n_iter=3,
            scoring='roc_auc',
            cv=inner_cv,
            random_state=42,
            n_jobs=1,
            refit=True
        )

        search.fit(X.iloc[tr_idx], y[tr_idx])

        predictions[val_idx] = (
            search.best_estimator_.predict_proba(
                X.iloc[val_idx]
            )[:, 1]
        )

        score = roc_auc_score(
            y[val_idx], predictions[val_idx]
        )

        fold_scores.append(score)
        best_params.append(search.best_params_)

        print(f"Fold {fold}: AUC = {score:.4f}")

    model_oof[name] = predictions
    best_parameters[name] = best_params

    print(
        "Mean fold AUC:",
        round(np.mean(fold_scores), 4)
    )

print("\nAll models trained and validated.")
```

```python

# CELL 6: MODEL EVALUATION AND VISUALIZATION

results = []

for name, probabilities in model_oof.items():

    pred = (probabilities >= 0.5).astype(int)

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y, pred),
        'Precision': precision_score(y, pred),
        'Recall': recall_score(y, pred),
        'F1 Score': f1_score(y, pred),
        'ROC AUC': roc_auc_score(y, probabilities)
    })

results_df = pd.DataFrame(results).sort_values(
    'ROC AUC', ascending=False
).reset_index(drop=True)

display(results_df.round(4))

# Five confusion matrices
fig, axes = plt.subplots(2, 3, figsize=(17,11))
axes = axes.flatten()

for i, (name, probabilities) in enumerate(model_oof.items()):

    pred = (probabilities >= 0.5).astype(int)

    sns.heatmap(
        confusion_matrix(y, pred),
        annot=True,
        fmt='d',
        cmap='Blues',
        ax=axes[i],
        xticklabels=['No', 'Yes'],
        yticklabels=['No', 'Yes']
    )

    axes[i].set_title(name)

axes[5].axis('off')
plt.tight_layout()
plt.show()

# ROC curves
plt.figure(figsize=(9,7))

for name, probabilities in model_oof.items():
    fpr, tpr, _ = roc_curve(y, probabilities)
    auc = roc_auc_score(y, probabilities)

    plt.plot(
        fpr, tpr,
        label=f'{name}: {auc:.4f}'
    )

plt.plot([0,1], [0,1], '--', color='gray')
plt.title('ROC Curve Comparison')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.show()

# Performance metrics
results_df.set_index('Model')[
    ['Accuracy','Precision','Recall','F1 Score','ROC AUC']
].plot(
    kind='bar',
    figsize=(13,6),
    ylim=(0,1),
    colormap='viridis'
)

plt.title('Five Model Performance Comparison')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

# Calibration curve of each model
plt.figure(figsize=(9,7))

for name, probabilities in model_oof.items():
    true_prob, pred_prob = calibration_curve(
        y, probabilities,
        n_bins=10,
        strategy='quantile'
    )

    plt.plot(
        pred_prob, true_prob,
        marker='o', label=name
    )

plt.plot([0,1], [0,1], '--', color='gray')
plt.title('Calibration Curve Comparison')
plt.xlabel('Predicted Probability')
plt.ylabel('Observed Positive Rate')
plt.legend()
plt.show()

# Probability distributions
plt.figure(figsize=(10,6))
sns.histplot(
    x=model_oof[results_df.iloc[0]['Model']],
    hue=y,
    bins=30,
    stat='density',
    common_norm=False,
    element='step'
)
plt.title('Best Model Probability Distribution')
plt.xlabel('Predicted Vaccination Probability')
plt.show()

print("\nCLASSIFICATION REPORTS")

for name, probabilities in model_oof.items():
    print(f"\n{name}")
    print(classification_report(
        y, (probabilities >= 0.5).astype(int)
    ))
```

```python

# CELL 7: COST-BASED EVALUATION

C_V = 300.0
C_H = 18500.0

P_HIGH_UNVAX = 0.041
P_HIGH_VAX = 0.0205

P_LOW_UNVAX = 0.011
P_LOW_VAX = 0.0055

def get_risk(df):

    age_high = (
        df['age_group'].astype(str) == '65+ Years'
    )

    chronic = pd.to_numeric(
        df['number_chronic_med_conditions'],
        errors='coerce'
    ).fillna(0)

    return (age_high | (chronic > 0)).astype(int).values


def calculate_expected_cost(probabilities, risk):

    total = 0
    offers = 0

    for p, hr in zip(probabilities, risk):

        if hr:
            pu, pv = P_HIGH_UNVAX, P_HIGH_VAX
        else:
            pu, pv = P_LOW_UNVAX, P_LOW_VAX

        offer_cost = (
            p * (C_V + pv*C_H)
            + (1-p)*pu*C_H
        )

        no_offer_cost = (
            p * (C_V + pu*C_H)
            + (1-p)*pu*C_H
        )

        if offer_cost < no_offer_cost:
            total += offer_cost
            offers += 1
        else:
            total += no_offer_cost

    return total, offers


risk = get_risk(train_df)

cost_results = []

for name, probabilities in model_oof.items():

    total_cost, offers = calculate_expected_cost(
        probabilities, risk
    )

    cost_results.append({
        'Model': name,
        'Expected Total Cost': total_cost,
        'Average Cost': total_cost / len(y),
        'Vaccine Offers': offers
    })

cost_df = pd.DataFrame(cost_results)

display(cost_df.sort_values(
    'Expected Total Cost'
))

plt.figure(figsize=(10,6))
sns.barplot(
    data=cost_df,
    x='Model',
    y='Expected Total Cost',
    hue='Model',
    palette='viridis',
    legend=False
)
plt.title('Expected Municipal Cost Comparison')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

print(
    "Note: Cost estimates are based on supplied assumptions, "
    "not verified official hackathon scoring."
)
```
